### zeitkontinuierliche Darstellung (Differentialgleichungen):

$$\dot{x}(t) = A x(t) + B u(t)$$

$$y(t) = C x(t) + D u(t)$$

### Zeitdiskret (Differenzengleichungen):

$$x_{k+1} = A x_k + B u_k$$
$$y_k = C x_k + D u_k$$


x
Zustandsvektor (interne Variablen, z. B. Position, Geschwindigkeit)

u
Eingangsvektor (Steuergrößen, z. B. Kraft, Spannung)

y
Ausgangsvektor (Messwerte, z. B. gemessene Position)

A
Systemmatrix (beschreibt, wie der Zustand sich selbst beeinflusst)

B
Eingangsmatrix (beschreibt, wie Eingänge auf den Zustand wirken)

C
Ausgangsmatrix (beschreibt, wie der Zustand den Ausgang beeinflusst)

D
Direktübertragung vom Eingang zum Ausgang (meist 0)


## Kalman Filter

Du brauchst ein diskretes Zustandsraummodell:

$$x_{k+1} = A x_k + B u_k + w_k$$
$$z_k = H x_k + v_k$$

	•	x_k: Systemzustand (z. B. Position, Geschwindigkeit)
	•	u_k: Steuereingang (z. B. äußere Kraft)
	•	z_k: Messung (z. B. Positionssensor)
	•	A, B, H: bekannte Modellmatrizen → direkt aus der Zustandsraumdarstellung
	•	w_k, v_k: Prozess- bzw. Messrauschen (mit Kovarianzen Q, R)
